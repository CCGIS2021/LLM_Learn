```python
stickers = ["with","example","science"]
target = "thehat"
```

```python
from collections import Counter
all_choose = []
target_counter = Counter(target)
for each_choose in stickers:
    choose_counter = Counter(each_choose)
    all_choose.append(choose_counter)
```

```python
def have_solution(stickers,target):
    total_set = set(''.join(stickers))
    target_set = set(target)
    for i in target_set:
        if i not in total_set:
            return False    
    return True
```

```python
import numpy as np
from sko.PSO import PSO


# 问题参数
num_items = len(all_choose)
up_num = max(target_counter.values())

# 定义适应度函数
def fitness_function(x):

    x = np.round(x)  # 将粒子位置四舍五入为0或1
    all_can_choose_char = {}
    for i,v in enumerate(x):
        if v >0:
            for key,count_num in all_choose[i].items():
                if key not in all_can_choose_char:
                    all_can_choose_char[key] = 0
                all_can_choose_char[key] += int(count_num*v)
    # print(x,all_can_choose_char)

    for key,count_num in target_counter.items():
        if key not in all_can_choose_char or all_can_choose_char[key] < count_num:
            return 1000
    # print(sum(x))
    return sum(x)


# 初始化PSO
pso = PSO(func=fitness_function, dim=num_items, pop=30, max_iter=100, lb=0, ub=up_num, w=0.8, c1=2, c2=2)

# 运行PSO算法
best_x, best_value = pso.run()

# 输出结果
best_x = np.round(best_x)
print("最优选择：", best_x)
print("最优总价值：", best_value)
```
